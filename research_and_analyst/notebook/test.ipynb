```python
import sys, os
from pydantic import BaseModel, Field
from typing import List, Optional, Tuple, Dict, Any, TypedDict, Union
from langgraph.graph import StateGraph, START, END
from langchain_core.messages import BaseMessage, AIMessage, HumanMessage, SystemMessage
from langgraph.checkpoint.memory import MemorySaver


# Get project root directory
project_root = os.path.dirname(os.path.join(os.getcwd(), "../.."))
```

```python
class Analyst(BaseModel):
    name: str =Field(description="Name of the Analyst")
    role: str =Field(description="Role of the analyst in the context of the topic")
    affiliation: str =Field(description="Primary organization or company the analyst is associated with")
    description: str =Field(description="Brief description of the analyst's expertise, background, or perspective on the topic")

    @property
    def persona(self) -> str:
        return f"{self.name} is a {self.role} at {self.affiliation}. {self.description}"

    
```

```python
analyst = Analyst(
    name="sunny savita",
    role="genai eng",
    affiliation="AI Research LAB",
    description="I am genai developer as well as mentor"
    )
```

```python
print(analyst.persona)
```
